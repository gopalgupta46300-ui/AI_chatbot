```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

# Set the path to the specific file you want to load (e.g., 'Churn_Modelling.csv')
file_path = "Churn_Modelling.csv"

# Load the dataset using the updated function
df = kagglehub.dataset_load(
    KaggleDatasetAdapter.PANDAS,
    "devahuja2808/churn-modeling-dataset",
    file_path
)

df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## DIvide the dataset into independent and dependent features

```python
X = df.iloc[:, 3:13]
y = df.iloc[:, 13]
```

```python
X.head()
```

```python
y.head()
```

## To add Germany & Spain, Male as category in independent dataset from Gender and Geography

```python
# categorical feature Engineering
pd.get_dummies(X["Geography"]) # for understanding get_dummies
```

```python
# it drop the first column without rueduce observation
geography = pd.get_dummies(X['Geography'], drop_first=True)
```

```python
gender = pd.get_dummies(X['Gender'], drop_first=True)
```

```python
# remove the gender and geography
X = X.drop(["Gender", "Geography"], axis=1)
X
```

```python
# add germany, spain, male
X = pd.concat([X, geography, gender], axis=1)
X
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 0 )
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.preprocessing import StandardScaler # StandardScaler :- based on z-value, and data is mean arounded
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
X_train.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.layers import LeakyReLU, ELU, ReLU, PReLU
from tensorflow.keras.layers import Dropout # drop layer to avoid overfitting, remove the unnessesary nodes
```

```python
## lets initialize ANN
classifier = Sequential()
print("classifier object created")
```

```python
## input layer
classifier.add(Dense(units=11,activation='relu'))
print("input layer")
```

```python
# first hidden layer
classifier.add(Dense(units=7, activation='relu'))
classifier.add(Dropout (0.2))
```

```python
# second hidden layer
classifier.add(Dense(units = 6, activation = 'relu'))
classifier.add(Dropout(0.3))
```

```python
# Output layer
classifier.add(Dense(units=1, activation = 'sigmoid'))
```

```python
classifier.compile(optimizer = 'adam', loss = "binary_crossentropy", metrics=['accuracy'])
print("compilation is done")
```

```python
import tensorflow
opt = tensorflow.keras.optimizers.Adam(learning_rate = 0.01)
```

```python
import keras
callback = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    min_delta=0.0001,
    patience=20,
    verbose=1,
    mode="auto",
    baseline=None,
    restore_best_weights=False,
    start_from_epoch=0,
)
```

```python
model_history = classifier.fit(X_train, y_train, validation_split = 0.33, batch_size = 20, epochs = 1000, callbacks =[callback])
```

```python
model_history.history.keys()
```

```python
# predict the test values

y_pred = classifier.predict(X_test)
y_pred = (y_pred >= 0.5)
```

```python
from sklearn.metrics import accuracy_score
score = accuracy_score(y_pred, y_test)
score
```
